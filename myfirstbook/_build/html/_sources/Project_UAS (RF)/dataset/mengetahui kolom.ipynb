```python
import pandas as pd

df = pd.read_csv('BME_TEST_DEPLOY.csv')

print("Target SEBELUM konversi:")
print(df['target'].value_counts().sort_index())

target_mapping = {
    1: 'Begin',
    2: 'Middle',
    3: 'End'
}

df['target'] = df['target'].map(target_mapping)

print("\nTarget SETELAH konversi:")
print(df['target'].value_counts())

df.to_csv('BME_TEST_DEPLOY_WITH_LABELS.csv', index=False)
print("\nFile berhasil disimpan sebagai 'dataset/BME_TEST_DEPLOY_WITH_LABELS.csv'")

print("\nContoh 5 baris pertama:")
print(df[['att1', 'att2', 'att3', 'target']].head())
```

```python
import pandas as pd

df_deploy = pd.read_csv('BME_TEST_DEPLOY_WITH_LABELS.csv')

print(f"Shape sebelum filter: {df_deploy.shape}")
print(f"Kolom: {df_deploy.columns.tolist()}")

df_target_only = df_deploy[['target']]

print(f"\nShape setelah filter (hanya target): {df_target_only.shape}")
print(f"Kolom: {df_target_only.columns.tolist()}")

print(f"\nDistribusi target:")
print(df_target_only['target'].value_counts().sort_index())

df_target_only.to_csv('BME_TEST_DEPLOY_TARGET_ONLY.csv', index=False)
print("\nFile berhasil disimpan sebagai 'BME_TEST_DEPLOY_TARGET_ONLY.csv'")

print("\n10 baris pertama:")
print(df_target_only.head(10))
```
