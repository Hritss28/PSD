```python
from scipy.io import arff
import pandas as pd
from sklearn.model_selection import train_test_split

data_train, meta_train = arff.loadarff("BME_TRAIN.arff")
df_train = pd.DataFrame(data_train)
df_train["target"] = df_train["target"].str.decode("utf-8").astype(int)
print("Jumlah data TRAIN:", len(df_train))
df_train.to_csv("BME_TRAIN.csv", index=False)

data_test, meta_test = arff.loadarff("BME_TEST.arff")
df_test = pd.DataFrame(data_test)
df_test["target"] = df_test["target"].str.decode("utf-8").astype(int) 

print("Total data TEST (ARFF):", len(df_test))  



df_test_kecil, df_test_deploy = train_test_split(
    df_test,
    train_size=8,               
    random_state=42,
    stratify=df_test["target"]    
)

print("Test model:", len(df_test_kecil))    
print("Test deploy:", len(df_test_deploy))           

df_test_kecil.to_csv("BME_TEST.csv", index=False)
df_test_deploy.to_csv("BME_TEST_DEPLOY.csv", index=False)

```
